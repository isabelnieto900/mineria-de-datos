```python
import zipfile
import os

# Nombre del archivo original
archivo_ig = "DoubleElectron_Run2012C_0.ig"
# Carpeta donde guardaremos los archivos extraídos
carpeta_destino = "archivos_extraidos_cern"

# Crear la carpeta si no existe
if not os.path.exists(carpeta_destino):
    os.makedirs(carpeta_destino)

try:
    # Python puede abrirlo como zip aunque la extensión sea .ig
    with zipfile.ZipFile(archivo_ig, 'r') as zip_ref:
        print(f"Descomprimiendo {archivo_ig}...")
        zip_ref.extractall(carpeta_destino)
        
    print(f"¡Listo! Archivos extraídos en la carpeta: {carpeta_destino}")
    
    # Mostrar qué encontramos dentro
    print("\nArchivos encontrados:")
    for root, dirs, files in os.walk(carpeta_destino):
        for file in files:
            print(f"- {os.path.join(root, file)}")

except zipfile.BadZipFile:
    print("Error: El archivo no parece ser un ZIP válido o está corrupto.")
except FileNotFoundError:
    print(f"Error: No encuentro el archivo '{archivo_ig}'. Asegúrate de que esté en la carpeta.")
```
